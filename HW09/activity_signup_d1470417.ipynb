```python
# 測試三：兩場去重後完全相同

# ==========================================
# HW09 兩場活動報名整理器（測試三）
# ==========================================

activity_names = ("桌遊體驗", "電影欣賞")
board_raw = " Amy, ben, amy, Cora "
movie_raw = " AMY, ben, CORA, amy "
unit_fees = {"桌遊體驗": 120, "電影欣賞": 80}

board_activity = activity_names[0]
movie_activity = activity_names[1]

board_list = board_raw.replace(" ", "").lower().split(",")
movie_list = movie_raw.replace(" ", "").lower().split(",")

board_set = set(board_list)
movie_set = set(movie_list)

both_set = board_set & movie_set
all_set = board_set | movie_set
board_only_set = board_set - movie_set
movie_only_set = movie_set - board_set

board_unique_sorted = sorted(board_set)
movie_unique_sorted = sorted(movie_set)
both_sorted = sorted(both_set)
board_only_sorted = sorted(board_only_set)
movie_only_sorted = sorted(movie_only_set)
all_sorted = sorted(all_set)
preview_top2 = all_sorted[:2]

board_count = len(board_set)
movie_count = len(movie_set)
both_count = len(both_set)
unique_total_count = len(all_set)
signup_total_count = board_count + movie_count

board_fee = board_count * unit_fees[board_activity]
movie_fee = movie_count * unit_fees[movie_activity]
total_fee = board_fee + movie_fee

summary = {
    "board_count": board_count,
    "movie_count": movie_count,
    "both_count": both_count,
    "unique_total_count": unique_total_count,
    "signup_total_count": signup_total_count,
    "board_fee": board_fee,
    "movie_fee": movie_fee,
    "total_fee": total_fee,
}

print("HW09 活動報名摘要")
print(f"活動：{board_activity}，每人 {unit_fees[board_activity]:.2f} 元")
print(f"活動：{movie_activity}，每人 {unit_fees[movie_activity]:.2f} 元")
print(f"桌遊整理串列：{board_list}")
print(f"電影整理串列：{movie_list}")
print(f"桌遊去重名單：{board_unique_sorted}")
print(f"電影去重名單：{movie_unique_sorted}")
print(f"桌遊人數：{summary['board_count']} 人")
print(f"電影人數：{summary['movie_count']} 人")
print(f"共同參加者：{both_sorted}")
print(f"共同人數：{summary['both_count']} 人")
print(f"只參加桌遊者：{board_only_sorted}")
print(f"只參加電影者：{movie_only_sorted}")
print(f"全部參加者：{all_sorted}")
print(f"總不重複人數：{summary['unique_total_count']} 人")
print(f"報名人次：{summary['signup_total_count']} 人次")
print(f"通知名單前兩位：{preview_top2}")
print(f"桌遊應收：{summary['board_fee']:.2f} 元")
print(f"電影應收：{summary['movie_fee']:.2f} 元")
print(f"總應收：{summary['total_fee']:.2f} 元")
```
